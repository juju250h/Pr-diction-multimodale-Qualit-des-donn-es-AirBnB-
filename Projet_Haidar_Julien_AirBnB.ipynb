```python
#Projet 1 : Data Cleaning & Processing
#Etant un exercice subjectif, voici notre proposition pour réparer le dataset


#Nettoyage des données

#Récupération de la liste des attributs
cols = list(data_set.columns)



#RÉPARATION DES DÉCALAGES (selon nous)


#Problème 1: La colonne 'elevator' présente des valeurs incohérentes
#Il faut détecter les cases n'ayant pas "yes" ou "no" ou NaN. Le mask contiendra True pour ces cases incorrectes.
mask_shift_1 = ~data_set['elevator'].isin(['yes', 'no']) & data_set['elevator'].notna()
idx_elevator = cols.index('elevator')
cols_shift_1 = cols[idx_elevator:]

#Mettre les colonnes en type object pour éviter les warnings
data_set[cols_shift_1] = data_set[cols_shift_1].astype(object)

#On décale d'une case vers la droite, juste à partir de l'ascenseur
data_set.loc[mask_shift_1, cols_shift_1[1:]] = data_set.loc[mask_shift_1, cols_shift_1[:-1]].values
#On met NaN pour les cases vides créées
data_set.loc[mask_shift_1, 'elevator'] = np.nan

#Problème 2: La colonne 'nb_parking' a sauté (13 lignes)
mask_shift_2 = data_set['Unnamed: 32'].isna()
idx_parking = cols.index('nb_parking')
cols_shift_2 = cols[idx_parking:]

#Mettre les colonnes en type object pour éviter les warnings
data_set[cols_shift_2] = data_set[cols_shift_2].astype(object)

#On décale d'une case vers la droite, juste à partir du parking
#Ainsi, on aura des valeurs de surface et price_per_night cohérentes
data_set.loc[mask_shift_2, cols_shift_2[1:]] = data_set.loc[mask_shift_2, cols_shift_2[:-1]].values
#On met NaN pour les cases vides créées
data_set.loc[mask_shift_2, 'nb_parking'] = np.nan


#Afficher un aperçu pour vérifier que tout est rentré dans l'ordre
data_set.head(10)
```

```python
######################## cancelling_conditions (3 Nan) ########################
#Ici, on va choisir la valeur la plus populaire
#Afficher les 3 annonces en question
print("--- Les 3 annonces sans conditions d'annulation ---")
lignes_vides_canc = data_set[data_set['cancelling_conditions'].isna()]
print(lignes_vides_canc[['type1', 'type2', 'price_per_night', 'cancelling_conditions']])

#Regarder la tendance des conditions selon le type de logement
print("\n--- Tendance des conditions par type (House vs Appartment) ---")
print(data_set.groupby('type2')['cancelling_conditions'].value_counts())

#Que ce soit pour appart ou maison, moderate est le grand gagnant
data_set['cancelling_conditions'] = data_set['cancelling_conditions'].fillna('moderated')
print(data_set.loc[[15, 30, 42], ['type1', 'type2', 'price_per_night', 'cancelling_conditions']])
```

```python
######################## animals_allowed / smoking_allowed / internet_access / TV ########################
#Ici c'est simple, si ce n'est pas renseigné c'est qu'il n'y a pas l'option

colonnes_binaires = ['animals_allowed', 'smoking_allowed', 'internet_access', 'TV']
# Pour chaque colonne, on remplace les NaN par 'no'
for col in colonnes_binaires:
    data_set[col] = data_set[col].fillna('no')

######################## Heating/ Air_conditioning / Washing_Machine / adapted_for_disabled ########################
#idem, si ce n'est pas renseigné c'est qu'il n'y a pas l'option
#Ajout des nouvelles colonnes à notre logique de "no" par défaut
colonnes_equipements_suite = ['Heating', 'Air_conditioning', 'Washing_Machine', 'adapted_for_disabled']

for col in colonnes_equipements_suite:
    data_set[col] = data_set[col].fillna('no')

######################## Elevator ########################
#idem, si ce n'est pas renseigné c'est qu'il n'y a pas l'option
data_set['elevator'] = data_set['elevator'].fillna('no')
```

```python
######################## nb_rooms / nb_beds / nb_bathrooms / nb_kitchen ########################
colonnes_num = ['nb_rooms', 'nb_beds', 'nb_bathrooms', 'nb_kitchen']
lignes_a_inspecter = data_set[data_set[colonnes_num].isna().any(axis=1)]

#Afficher les informations pertinentes pour l'analyse
print("--- Analyse des descriptions ---")
print(lignes_a_inspecter[['ID', 'type2', 'nb_max_people', 'nb_rooms', 'nb_beds', 'nb_bathrooms', 'nb_kitchen', 'french_description']])

#On remplit avec les données les pertinentes à partir des informations obtenues par le print
#Remplissage des chambres (nb_rooms)
data_set.loc[19, 'nb_rooms'] = 2 
data_set.loc[44, 'nb_rooms'] = 1 
data_set.loc[80, 'nb_rooms'] = 1
data_set.loc[98, 'nb_rooms'] = 4

#Remplissage des lits (nb_beds)
data_set.loc[51, 'nb_beds'] = 1
data_set.loc[58, 'nb_beds'] = 1

#Remplissage des salles de bain (nb_bathrooms)
data_set.loc[27, 'nb_bathrooms'] = 1
data_set.loc[57, 'nb_bathrooms'] = 1
data_set.loc[87, 'nb_bathrooms'] = 1

#Remplissage de la cuisine (nb_kitchen)
data_set.loc[71, 'nb_kitchen'] = 1

#Affichage des modifications
index_modifies = [19, 27, 44, 51, 57, 58, 71, 80, 87, 98]
print("--- Visualisation ---")
print(data_set.loc[index_modifies, ['ID', 'nb_rooms', 'nb_beds', 'nb_bathrooms', 'nb_kitchen', 'french_description']])
```

```python
######################## price_per_night (3 Nan) ########################
#Puisque c'est ce que l'IA doit prédire, on a décidé de supprimer ces trois lignes pour moins "fausser" l'IA
#Il n'en restera plus que 97
data_set['price_per_night'] = pd.to_numeric(data_set['price_per_night'], errors='coerce')
data_set = data_set.dropna(subset=['price_per_night']).copy()

######################## 'nb_parking / 'reduction_per_week', 'reduction_per_month', 'cleaning_cost'] ########################
#Eviter les warnings
cols_derniers_num = ['nb_parking', 'reduction_per_week', 'reduction_per_month', 'cleaning_cost']
for col in cols_derniers_num:
    data_set[col] = pd.to_numeric(data_set[col], errors='coerce')

#Si l'option n'est pas renseigné, c'est qu'il n'y en pas
data_set['nb_parking'] = data_set['nb_parking'].fillna(0)
data_set['reduction_per_week'] = data_set['reduction_per_week'].fillna(0)
data_set['reduction_per_month'] = data_set['reduction_per_month'].fillna(0)

#On remplit par la médiane pour les frais de ménage puisqu'il y en a toujours des frais
data_set['cleaning_cost'] = data_set['cleaning_cost'].fillna(data_set['cleaning_cost'].median())

print("État final des NaN")
print(data_set.isna().sum().sum())

data_set.head(6)
```

```python
############ Prétraitement ############ 

#Encodage des variables catégorielles

#Label encoding avec "yes" <-> 1 et "no" <-> 0
colonnes_binaires = ['animals_allowed', 'smoking_allowed', 'internet_access', 'TV', 
                     'Heating', 'Air_conditioning', 'Washing_Machine', 'Dishwasher', 
                     'adapted_for_disabled', 'elevator']
for col in colonnes_binaires:
    data_set[col] = data_set[col].map({'yes': 1, 'no': 0})

#One-Hot Encoding
#Chaque attribut présentant plusieurs options est transformé en plusieurs sous attribut avec des "no" <-> 0 et des "yes" <-> 1
colonnes_cat = ['type1', 'type2', 'cancelling_conditions']

for col in colonnes_cat:
    index_position = data_set.columns.get_loc(col)
    dummies = pd.get_dummies(data_set[col], prefix=col, dtype=int)
    data_set = data_set.drop(col, axis=1)
    for i, nouvelle_colonne in enumerate(dummies.columns):
        data_set.insert(index_position + i, nouvelle_colonne, dummies[nouvelle_colonne])

# Mise à l’échelle des variables (sauf pour l'attribut cible : price_per_night)
from sklearn.preprocessing import StandardScaler

colonnes_num_a_scaler = ['nb_max_people', 'nb_rooms', 'nb_beds', 'nb_bathrooms', 
                         'nb_kitchen', 'nb_parking', 'surface', 
                         'reduction_per_week', 'reduction_per_month', 'cleaning_cost']
scaler = StandardScaler()
data_set[colonnes_num_a_scaler] = scaler.fit_transform(data_set[colonnes_num_a_scaler])
```

```python
#Projet 2: Premier modèle de machine learning
#Ici, on fait l'hypothèse de n'utiliser que les attributs numériques, catégoriels et booléens
#La régression linéaire est le point de comparaison par défaut avec un autre modèle.
#Ici, on va utiliser Random Forest et donc le comparer avec la régression linéaire.

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
```

```python
#Cross validation
from sklearn.model_selection import KFold, cross_validate
from sklearn.metrics import make_scorer
from sklearn.ensemble import RandomForestRegressor, GradientBoostingRegressor

#Variable d'intérêt
y_global = data_set['price_per_night']

#Attributs à éliminer
colonnes_texte = ['price_per_night', 'ID', 'URL', 'title', 'french_description', 
                  'english_description', 'location', 'Unnamed: 30', 'Unnamed: 31', 'Unnamed: 32']
X_global = data_set.drop(columns=[col for col in colonnes_texte if col in data_set.columns])


#Le modèle donne une réponse correcte que si le résultat prédit à 10% d'écart avec le vrai prix
def calcul_accuracy_10_pourcent(y_vrai, y_predit):
    ecart = np.abs((y_vrai - y_predit) / y_vrai)
    return np.mean(ecart <= 0.10) * 100

accuracy_scorer = make_scorer(calcul_accuracy_10_pourcent)

#Les indicateurs que l'on souhaite analyser
les_metriques = {
    'R2': 'r2',
    'MAE': 'neg_mean_absolute_error',
    'MSE': 'neg_mean_squared_error',
    'Accuracy': accuracy_scorer
}

#Instanciation des modèles
lr_model = LinearRegression()
rf_model = RandomForestRegressor(n_estimators=100, random_state=42)
gb_model = GradientBoostingRegressor(n_estimators=100, random_state=42)

#Cross validation avec un découpage par 10 (aléatoirement)
print("Calcul de la validation croisée sur 10 segments en cours...\n")

#Découpage du dataset aléatoirement
decoupage = KFold(n_splits=10, shuffle=True) #L'aléatoire est activé.
resultats_lr = cross_validate(lr_model, X_global, y_global, cv=decoupage, scoring=les_metriques)
resultats_rf = cross_validate(rf_model, X_global, y_global, cv=decoupage, scoring=les_metriques)
resultats_gb = cross_validate(gb_model, X_global, y_global, cv=decoupage, scoring=les_metriques)
#Affichage des résultats
print("\n--- RÉSULTATS DE LA RÉGRESSION LINEAIRE SIMPLE ---")

#RÉGRESSION LINÉAIRE
print("RÉGRESSION LINÉAIRE :")
print(f"  -> MAE par pli : {np.array2string(-resultats_lr['test_MAE'], precision=1)}")
print(f"  -> MSE par pli : {np.array2string(-resultats_lr['test_MSE'], precision=1)}")
print(f"  -> Acc par pli : {np.array2string(resultats_lr['test_Accuracy'], precision=1)} %")
print(f"  --> MOYENNE : MAE = {-resultats_lr['test_MAE'].mean():.2f} € | MSE = {-resultats_lr['test_MSE'].mean():.2f} | R² = {resultats_lr['test_R2'].mean():.4f} | Précision = {resultats_lr['test_Accuracy'].mean():.2f} %\n")

#RANDOM FOREST
print("RANDOM FOREST :")
print(f"  -> MAE par pli : {np.array2string(-resultats_rf['test_MAE'], precision=1)}")
print(f"  -> MSE par pli : {np.array2string(-resultats_rf['test_MSE'], precision=1)}")
print(f"  -> Acc par pli : {np.array2string(resultats_rf['test_Accuracy'], precision=1)} %")
print(f"  --> MOYENNE : MAE = {-resultats_rf['test_MAE'].mean():.2f} € | MSE = {-resultats_rf['test_MSE'].mean():.2f} | R² = {resultats_rf['test_R2'].mean():.4f} | Précision = {resultats_rf['test_Accuracy'].mean():.2f} %\n")

#LE GRADIENT BOOSTING
print("GRADIENT BOOSTING :")
print(f"  -> MAE par pli : {np.array2string(-resultats_gb['test_MAE'], precision=1)}")
print(f"  -> MSE par pli : {np.array2string(-resultats_gb['test_MSE'], precision=1)}")
print(f"  -> Acc par pli : {np.array2string(resultats_gb['test_Accuracy'], precision=1)} %")
print(f"  --> MOYENNE : MAE = {-resultats_gb['test_MAE'].mean():.2f} € | MSE = {-resultats_gb['test_MSE'].mean():.2f} | R² = {resultats_gb['test_R2'].mean():.4f} | Précision = {resultats_gb['test_Accuracy'].mean():.2f} %")
```

Ici, on étudie la régression linéaire car c'est la base. C'est un modèle simple et rapide à réaliser. De plus, si on remarque que la régression linéaire fournit des performances semblables à celles d'un autre modèle plus complexe (random forest, etc.), alors c'est la régression qui est conservée car elle est moins coûteuse en industrie, par exemple.

Pour comparer la régression linéaire avec le random forest et le gradient boosting, on utilise le concept de cross-validation. Ainsi, le dataset sert de données d'entraînement et d'évaluation pour les modèles. La segmentation du dataset se fait en 10 blocs (folds) de façon aléatoire.

L'accuracy a été définie comme la proportion de réponses ayant un écart max de 10 % avec le prix réel.

Pour une graine aléatoire, on peut déjà observer que les deux modèles ont une précision similaire. Pour la régression linéaire, on a 50,22 % et pour le random forest, c'est 60 %.

Mais pour analyser d'une meilleure façon les résultats, mieux vaut comparer les indicateurs sur plusieurs graines aléatoires.

```python
#Analyse de 20 seeds différents
print("\nCalcul des 4 métriques sur 20 découpages aléatoires (seeds)...")

graines = range(1, 21)

#On crée des dictionnaires pour stocker l'historique des 4 métriques
historique_lr = {'MAE': [], 'MSE': [], 'R2': [], 'Accuracy': []}
historique_rf = {'MAE': [], 'MSE': [], 'R2': [], 'Accuracy': []}
historique_gb = {'MAE': [], 'MSE': [], 'R2': [], 'Accuracy': []}

for seed in graines:
    #Découpage avec le nouveau seed
    decoupage_test = KFold(n_splits=10, shuffle=True, random_state=seed)
    
    #Évaluation de la Régression Linéaire avec les 4 métriques
    res_lr = cross_validate(lr_model, X_global, y_global, cv=decoupage_test, scoring=les_metriques)
    historique_lr['MAE'].append(-res_lr['test_MAE'].mean())
    historique_lr['MSE'].append(-res_lr['test_MSE'].mean())
    historique_lr['R2'].append(res_lr['test_R2'].mean())
    historique_lr['Accuracy'].append(res_lr['test_Accuracy'].mean())
    
    #Évaluation du Random Forest avec les 4 métriques
    res_rf = cross_validate(rf_model, X_global, y_global, cv=decoupage_test, scoring=les_metriques)
    historique_rf['MAE'].append(-res_rf['test_MAE'].mean())
    historique_rf['MSE'].append(-res_rf['test_MSE'].mean())
    historique_rf['R2'].append(res_rf['test_R2'].mean())
    historique_rf['Accuracy'].append(res_rf['test_Accuracy'].mean())

    #Évaluation du Gradient Boosting avec les 4 métriques
    res_gb = cross_validate(gb_model, X_global, y_global, cv=decoupage_test, scoring=les_metriques)
    historique_gb['MAE'].append(-res_gb['test_MAE'].mean())
    historique_gb['MSE'].append(-res_gb['test_MSE'].mean())
    historique_gb['R2'].append(res_gb['test_R2'].mean())
    historique_gb['Accuracy'].append(res_gb['test_Accuracy'].mean())


fig, axs = plt.subplots(2, 2, figsize=(16, 10))
fig.suptitle('Stabilité des 3 modèles selon le découpage aléatoire (20 Seeds)', fontsize=18, fontweight='bold')

#Graphique MAE (Haut Gauche)
axs[0, 0].plot(graines, historique_lr['MAE'], marker='o', color='blue', label='Régression Linéaire', linestyle=':')
axs[0, 0].plot(graines, historique_rf['MAE'], marker='s', color='green', label='Random Forest', linestyle='--')
axs[0, 0].plot(graines, historique_gb['MAE'], marker='^', color='red', label='Gradient Boosting', linestyle='-') # LIGNE AJOUTÉE
axs[0, 0].set_title('1. Erreur Absolue Moyenne (MAE) - Plus bas = Mieux', fontsize=12)
axs[0, 0].set_ylabel('Euros (€)')
axs[0, 0].grid(True, linestyle=':', alpha=0.7)
axs[0, 0].set_xticks(graines)

#Graphique MSE (Haut Droite)
axs[0, 1].plot(graines, historique_lr['MSE'], marker='o', color='blue', linestyle=':')
axs[0, 1].plot(graines, historique_rf['MSE'], marker='s', color='green', linestyle='--')
axs[0, 1].plot(graines, historique_gb['MSE'], marker='^', color='red', linestyle='-') # LIGNE AJOUTÉE
axs[0, 1].set_title('2. Erreur Quadratique (MSE) - Sensibilité aux extrêmes', fontsize=12)
axs[0, 1].grid(True, linestyle=':', alpha=0.7)
axs[0, 1].set_xticks(graines)

#Graphique R² (Bas Gauche)
axs[1, 0].plot(graines, historique_lr['R2'], marker='o', color='blue', linestyle=':')
axs[1, 0].plot(graines, historique_rf['R2'], marker='s', color='green', linestyle='--')
axs[1, 0].plot(graines, historique_gb['R2'], marker='^', color='red', linestyle='-') # LIGNE AJOUTÉE
axs[1, 0].set_title('3. Score R² - Plus haut = Mieux', fontsize=12)
axs[1, 0].set_xlabel('Numéro du Seed')
axs[1, 0].grid(True, linestyle=':', alpha=0.7)
axs[1, 0].set_xticks(graines)

#Graphique Accuracy (Bas Droite)
axs[1, 1].plot(graines, historique_lr['Accuracy'], marker='o', color='blue', linestyle=':')
axs[1, 1].plot(graines, historique_rf['Accuracy'], marker='s', color='green', linestyle='--')
axs[1, 1].plot(graines, historique_gb['Accuracy'], marker='^', color='red', linestyle='-') # LIGNE AJOUTÉE
axs[1, 1].set_title('4. Précision à ±10% - Plus haut = Mieux', fontsize=12)
axs[1, 1].set_xlabel('Numéro du Seed')
axs[1, 1].set_ylabel('Pourcentage (%)')
axs[1, 1].grid(True, linestyle=':', alpha=0.7)
axs[1, 1].set_xticks(graines)

handles, labels = axs[0, 0].get_legend_handles_labels()
fig.legend(handles, labels, loc='upper right', bbox_to_anchor=(0.98, 0.95), fontsize=12)

plt.tight_layout(rect=[0, 0.03, 1, 0.95])
plt.show()
```

```python
#Prétraitement du texte
nltk.download('stopwords', quiet=True)

stop_words_fr = set(stopwords.words('french'))
stop_words_en = set(stopwords.words('english'))
#On rajoute des mots peu intéressants
tous_les_stopwords = stop_words_fr | stop_words_en | {
    'plus', 'tout', 'avec', 'les', 'des', 'très', 'cet', 'cette', 'aussi'
}

#Chargement des deux modèles spaCy (un par langue)
nlp_fr = spacy.load('fr_core_news_sm')
nlp_en = spacy.load('en_core_web_sm')

def pretraitement_nlp(texte, nlp):
    if not isinstance(texte, str):
        return ""
    #On met tout en minuscule
    texte = texte.lower()
    #Elimination de tout ce qui n'est pas alphanumérique
    texte = re.sub(r'[^\w\s]', ' ', texte)
    #Lemmatisation avec spaCy (comprend la langue, pas de coupe aveugle)
    doc = nlp(texte)
    tokens_propres = [
        token.lemma_ for token in doc
        if token.text not in tous_les_stopwords
        and not token.is_space
        and len(token.text) > 2  #on filtre les mots trop courts
    ]
    return " ".join(tokens_propres)

#Traitement séparé FR et EN avec le bon modèle pour chaque langue
print("Prétraitement du texte en cours...")
titre_propre = data_set['title'].fillna('').apply(lambda t: pretraitement_nlp(t, nlp_fr))
texte_fr     = data_set['french_description'].fillna('').apply(lambda t: pretraitement_nlp(t, nlp_fr))
texte_en     = data_set['english_description'].fillna('').apply(lambda t: pretraitement_nlp(t, nlp_en))

#Fusion après lemmatisation dans la bonne langue
texte_propre = titre_propre + " " + texte_fr + " " + texte_en
```

```python
#Prétraitement des images
#les images seront mise en noir et blanc et de taille 64*64.
#Cela réduira le temps de traitement du réseau de neurone qui se chargera de créer le vecteur embedding des images.
#Pour éviter de davantage "aloudir" les vecteurs inputs, le vecteur embedding fourni sera de petite taille


#Chaque image => noir & blanc => 64x64 => normalisée entre 0 et 1
#3 images par logement => 10 vecteurs embedding de taille de 10
#Bien sûr, tout ce traitement résulte en une perte d'information très grande.
#Cependant, c'est le prix à payer pour éviter d'inutilement alourdir l'input.

from PIL import Image


TAILLE = (64, 64)
COLONNES_IMAGES = ['Unnamed: 30', 'Unnamed: 31', 'Unnamed: 32']

def charger_image(chemin):
    try:
        img = Image.open(chemin).convert('L')  #Noir et Blanc
        img = img.resize(TAILLE)
        return np.array(img) / 255.0 #Normalisation pour avoir la même échelle pour toutes les images pour l'entrainement du CNN.
    except Exception:
        return np.zeros(TAILLE)  #Image noire si manquante

print("Chargement et prétraitement des images...")
X_images = []

for idx, row in data_set.iterrows():
    images_logement = []
    for col in COLONNES_IMAGES:
        chemin = row[col]
        if isinstance(chemin, str):
            img = charger_image(chemin)
        else:
            img = np.zeros(TAILLE)
        images_logement.append(img)
    X_images.append(images_logement)

X_images = np.array(X_images)          #shape: (97, 3, 64, 64)
X_images = X_images[..., np.newaxis]   #shape: (97, 3, 64, 64, 1)
print(f"Forme du tableau images : {X_images.shape}")

#Affichage des 3 images d'un logement exemple
fig, axes = plt.subplots(1, 3, figsize=(10, 3))
for i in range(3):
    axes[i].imshow(X_images[0, i, :, :, 0], cmap='gray')
    axes[i].set_title(f'Image {i+1}')
    axes[i].axis('off')
plt.suptitle("Les 3 images d'un logement après prétraitement", fontweight='bold')
plt.tight_layout()
plt.show()
```

```python
#Construction du CNN avec Keras
#Il y a 16 filtres, puis 32 filtres puis 64 neurones, puis 10 neurones (embedding) et enfin 1 neurone (prix)

import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
from tensorflow.keras.callbacks import EarlyStopping

#Sous-modèle CNN partagé entre les 3 images
def construire_cnn_partage():
    entree = keras.Input(shape=(64, 64, 1))
    x = layers.Conv2D(16, (3, 3), activation='relu', padding='same')(entree)
    x = layers.MaxPooling2D((2, 2))(x)
    x = layers.Conv2D(32, (3, 3), activation='relu', padding='same')(x)
    x = layers.MaxPooling2D((2, 2))(x)
    x = layers.Flatten()(x)
    x = layers.Dense(64, activation='relu')(x)
    embedding = layers.Dense(10, activation='relu', name='embedding')(x)
    return keras.Model(inputs=entree, outputs=embedding)

cnn_partage = construire_cnn_partage()

#Modèle global: 3 entrées image => 3 embeddings => concaténation => prix
entree_img1 = keras.Input(shape=(64, 64, 1), name='image_1')
entree_img2 = keras.Input(shape=(64, 64, 1), name='image_2')
entree_img3 = keras.Input(shape=(64, 64, 1), name='image_3')

#Le même CNN est appliqué aux 3 images (poids partagés)
emb1 = cnn_partage(entree_img1)  
emb2 = cnn_partage(entree_img2)  
emb3 = cnn_partage(entree_img3)  

#Concaténation des 3 embeddings => vecteur de 30 dimensions
x = layers.Concatenate(name='embedding_final')([emb1, emb2, emb3])
sortie = layers.Dense(1, name='sortie')(x)

modele_cnn = keras.Model(
    inputs=[entree_img1, entree_img2, entree_img3],
    outputs=sortie
)
modele_cnn.summary()
```

```python
#Pour obtenir des vecteurs embeddings exploitables, il faut les orientés grâce à l'output (prix)
y_global_np = y_global.values

#Entraînement du CNN
modele_cnn.compile(
    optimizer='adam',
    loss='mean_squared_error',
    metrics=['mae']
)

historique_entrainement = modele_cnn.fit(
    [X_images[:, 0], X_images[:, 1], X_images[:, 2]],
    y_global_np,
    epochs=100,
    batch_size=16,
    validation_split=0.2, #20% des données utilisées pour le test
    verbose=1
)

#Courbe d'apprentissage
plt.figure(figsize=(10, 4))
plt.plot(historique_entrainement.history['mae'], label='MAE train')
plt.plot(historique_entrainement.history['val_mae'], label='MAE validation')
plt.title("Courbe d'apprentissage du CNN")
plt.xlabel("Époque")
plt.ylabel("MAE (€)")
plt.legend()
plt.grid(True, linestyle=':', alpha=0.7)
plt.tight_layout()
plt.show()
```

Dans ce graphe, on peut voir que le MAE_train descend progressivement jusqu'à atteindre 4.8973€. C'est un signe que le modèle apprend bien le data de l'entraînement. Cependant, on peut voir que MAE_validation, partant de 140.9683€, finit à 76.6679€. Donc tout ce que le modèle a fait était de l'overfitting. Il ne fait qu'apprendre par-coeur le set provenant de l'entraînement.
On décide quand même de garder ce réseau de neurone. Il n'y a pas assez de données pour que le réseau de neurone apprenne vraiment.

```python
#Modèle complexe
#Fonction qui convertit une input en une input utilisable pour le gradient boosting complexe (avec texte, image...)

def predire_prix(nouvel_exemple_brut, chemins_images):

    df_new = pd.DataFrame([nouvel_exemple_brut])

    df_new = df_new.drop(columns=['ID', 'URL', 'location'], errors='ignore')


    colonnes_binaires = ['animals_allowed', 'smoking_allowed', 'internet_access', 'TV',
                         'Heating', 'Air_conditioning', 'Washing_Machine', 'Dishwasher',
                         'adapted_for_disabled', 'elevator']
    for col in colonnes_binaires:
        if col in df_new.columns:
            df_new[col] = df_new[col].map({'yes': 1, 'no': 0})


    colonnes_cat = ['type1', 'type2', 'cancelling_conditions']
    for col in colonnes_cat:
        dummies = pd.get_dummies(df_new[col], prefix=col, dtype=int)
        df_new = df_new.drop(col, axis=1)
        for nouvelle_colonne in dummies.columns:
            df_new[nouvelle_colonne] = dummies[nouvelle_colonne]

    colonnes_ohe_attendues = [c for c in X_global.columns
                              if c.startswith(('type1_', 'type2_', 'cancelling_'))]
    colonnes_manquantes = {c: [0] for c in colonnes_ohe_attendues if c not in df_new.columns}
    if colonnes_manquantes:
        df_new = pd.concat([df_new, pd.DataFrame(colonnes_manquantes)], axis=1)


    colonnes_num_a_scaler = ['nb_max_people', 'nb_rooms', 'nb_beds', 'nb_bathrooms',
                             'nb_kitchen', 'nb_parking', 'surface',
                             'reduction_per_week', 'reduction_per_month', 'cleaning_cost']
    df_new[colonnes_num_a_scaler] = scaler.transform(df_new[colonnes_num_a_scaler])


    description_brute = (
        df_new['title'].fillna('') + " " +
        df_new['french_description'].fillna('') + " " +
        df_new['english_description'].fillna('')
    ).str.lower()

    for col, mots in mots_cles.items():
        for mot in mots:
            mask = (
                description_brute.str.contains(mot, na=False) &
                ~description_brute.str.contains(
                    f"pas d.{mot}|sans {mot}|no {mot}|without {mot}", na=False) &
                (df_new[col] == 0)
            )
            df_new.loc[mask, col] = 1


    titre_propre = df_new['title'].fillna('').apply(lambda t: pretraitement_nlp(t, nlp_fr))
    texte_fr_new = df_new['french_description'].fillna('').apply(lambda t: pretraitement_nlp(t, nlp_fr))
    texte_en_new = df_new['english_description'].fillna('').apply(lambda t: pretraitement_nlp(t, nlp_en))
    texte_propre_new = titre_propre + " " + texte_fr_new + " " + texte_en_new


    colonnes_a_exclure = ['price_per_night', 'title', 'french_description',
                          'english_description', 'Unnamed: 30', 'Unnamed: 31', 'Unnamed: 32']
    X_new = df_new.drop(columns=[c for c in colonnes_a_exclure if c in df_new.columns])

    bow_new = vectoriseur_bow.transform(texte_propre_new)
    df_bow_new = pd.DataFrame(
        bow_new.toarray(),
        columns=[f"bow_{m}" for m in vectoriseur_bow.get_feature_names_out()]
    )
    X_new_bow = pd.concat([X_new.reset_index(drop=True),
                           df_bow_new.reset_index(drop=True)], axis=1)


    images_logement = []
    for chemin in chemins_images:
        img = charger_image(chemin)
        images_logement.append(img)

    X_img_new = np.array(images_logement)[np.newaxis, ...]  
    X_img_new = X_img_new[..., np.newaxis]                  

    embed_img_new = extracteur_embedding.predict(
        [X_img_new[:, 0], X_img_new[:, 1], X_img_new[:, 2]], verbose=0
    )
    df_embed_new = pd.DataFrame(
        embed_img_new,
        columns=[f"img_emb_{i}" for i in range(embed_img_new.shape[1])]
    )

    X_final_new = pd.concat([X_new_bow.reset_index(drop=True),
                             df_embed_new.reset_index(drop=True)], axis=1)
    

    X_final_new = X_final_new[X_global_final.columns]
    
    print(f"Shape vecteur final : {X_final_new.shape}")
    
    prix_predit = gb_model.predict(X_final_new)[0]
    print(f"\nPrix prédit par nuit : {prix_predit:.2f} €")
    return prix_predit

# Entraînement final du modèle sur toutes les données
gb_model.fit(X_global_final, y_global)
```

```python
#Modèle simple
#Fonction qui convertit une input en une input utilisable pour le gradient boosting simple
def predire_prix_simple(nouvel_exemple_brut):

    df_new = pd.DataFrame([nouvel_exemple_brut])

    #ÉTAPE 1: Suppression ID, URL, Location
    df_new = df_new.drop(columns=['ID', 'URL', 'location'], errors='ignore')

    #ÉTAPE 2: Label Encoding yes/no → 0/1
    colonnes_binaires = ['animals_allowed', 'smoking_allowed', 'internet_access', 'TV',
                         'Heating', 'Air_conditioning', 'Washing_Machine', 'Dishwasher',
                         'adapted_for_disabled', 'elevator']
    for col in colonnes_binaires:
        if col in df_new.columns:
            df_new[col] = df_new[col].map({'yes': 1, 'no': 0})

    #ÉTAPE 3: One-Hot Encoding
    colonnes_cat = ['type1', 'type2', 'cancelling_conditions']
    for col in colonnes_cat:
        dummies = pd.get_dummies(df_new[col], prefix=col, dtype=int)
        df_new = df_new.drop(col, axis=1)
        for nouvelle_colonne in dummies.columns:
            df_new[nouvelle_colonne] = dummies[nouvelle_colonne]

    #Colonnes OHE manquantes → 0
    colonnes_ohe_attendues = [c for c in X_global.columns
                              if c.startswith(('type1_', 'type2_', 'cancelling_'))]
    colonnes_manquantes = {c: [0] for c in colonnes_ohe_attendues if c not in df_new.columns}
    if colonnes_manquantes:
        df_new = pd.concat([df_new, pd.DataFrame(colonnes_manquantes)], axis=1)

    #ÉTAPE 4: Normalisation
    colonnes_num_a_scaler = ['nb_max_people', 'nb_rooms', 'nb_beds', 'nb_bathrooms',
                             'nb_kitchen', 'nb_parking', 'surface',
                             'reduction_per_week', 'reduction_per_month', 'cleaning_cost']
    df_new[colonnes_num_a_scaler] = scaler.transform(df_new[colonnes_num_a_scaler])

    #ÉTAPE 5: Suppression colonnes inutiles
    colonnes_a_exclure = ['price_per_night', 'title', 'french_description',
                          'english_description', 'Unnamed: 30', 'Unnamed: 31', 'Unnamed: 32']
    X_new = df_new.drop(columns=[c for c in colonnes_a_exclure if c in df_new.columns])

    #Réalignement avec X_global
    X_new = X_new[X_global.columns]

    print(f"Shape vecteur final : {X_new.shape}") 

    prix_predit = gb_model_simple.predict(X_new)[0]
    print(f"\nPrix prédit par nuit : {prix_predit:.2f} €")
    return prix_predit


gb_model_simple = GradientBoostingRegressor(n_estimators=100, random_state=42)
gb_model_simple.fit(X_global, y_global)
```
